# Desenvolvimento de Rede Neural para porta XOR

````
import numpy as np
import matplotlib.pyplot as plt  #[Adicionado para plotar a curva de aprendizado nos testes]
````

## Implementação da MLP — Arquitetura fixa (2 entradas → 2 neurônios ocultos → 1 saída)

````
# A criação de uma classe permite não deixar as variáveis "despersas", com uma Classe, tudo que é pertencente a uma rede neural, fica nela
# EX: class MLP: | def train(self) pass | def preditct (self) pass, podendo ser criado uma instância da rede neural, logo: rede = MLP() | rede.train() | rede.predict()
class MLP:
  def __init__(self): #"__init__" é utilizado para iniciar atributos, nesse caso o ""self
    pass              # comando vazio para evitar erro na função que precisa ter alguma "ação"

  def train(self, inputs, outputs, alpha, epochs):
    #Parâmetros que a função train deverá receber
    #[Aqui o self é essencial para chamar os parâmetros fora da função]
    self.inputs = inputs
    self.outputs = outputs
    self.alpha = alpha   #Taxa de aprendizado do Algoritmo
    self.epochs = epochs #Quantidade de Iterações na etapa de treinamento

    #Definição dos Pesos e Bias (inicialização aleaória)
    w11 = np.random.uniform(0,1) #[np(chama o numpy).random(seta aleatoriedade).uniform(seta uma aleatoriedade "uniform" entre 0 e 1)]
    w12 = np.random.uniform(0,1)
    w21 = np.random.uniform(0,1)
    w22 = np.random.uniform(0,1)

    wh1 = np.random.uniform(0,1)
    wh2 = np.random.uniform(0,1)

    b1 = np.random.uniform(0,1)
    b2 = np.random.uniform(0,1)
    b3 = np.random.uniform(0,1)

    #LAÇOS DE REPETIÇÃO PARA TREINAMENTO
    for i in range(epochs):         #Iteração para a quantidade de épocas selecionadas
      for j in range(len(inputs)):  #"len(inputs)" retorna quantos exemplos de treinamento existem considerando todo o conjunto de dados(inputs)
                                    # Por exemplo: inputs = {[0,0] [0,1] [1,0] [1,1]}

    #Execução de cálculos que serão passados adiantes para a camada oculta
    #Utiliza a função de ativação sigmoid, que possibilita transformar qualquer valor em um número entre 0 e 1 --> f(x) = 1 / 1 + e⁻x
    # Então aplicando a o sigmoid temos: h = f(x1*w1 + x2*w2 + b1 ) -> em que o resultado do "treinamento" é convertido para um valor entre 0 e 1
        h1 = 1 / (1 + np.exp(-(w11 * inputs[j][0] + w12 * inputs[j][1] + b1)))
        h2 = 1 / (1 + np.exp(-(w21 * inputs[j][0] + w22 * inputs[j][1] + b2)))

        #Execução de cálculos que serão passados adiantes para a camada de saída
        y = 1 / (1 + np.exp(-(wh1 * h1 + wh2 * h2 + b3)))

        #Calculamos o erro
        error = outputs[j][0] - y # Access the scalar value from the list

        #Calculo de derivadas parciais
        derivative_y = y * (1 - y) * error
        derivative_h1 = h1 * (1 - h1) * wh1 * derivative_y
        derivative_h2 = h2 * (1 - h2) * wh2 * derivative_y

        #Calculo dos deltas que serão utilizados para atualização dos pesos
        delta_w11 = alpha * derivative_h1 * inputs[j][0]
        delta_w12 = alpha * derivative_h2 * inputs[j][0]
        delta_w21 = alpha * derivative_h1 * inputs[j][1]
        delta_w22 = alpha * derivative_h2 * inputs[j][1]

        delta_b1 = alpha * derivative_h1
        delta_b2 = alpha * derivative_h2
        delta_b3 = alpha * derivative_y

        delta_wh1 = alpha * derivative_y * h1
        delta_wh2 = alpha * derivative_y * h2

        #Atualização dos pesos
        w11 += delta_w11
        w12 += delta_w12
        w21 += delta_w21
        w22 += delta_w22

        wh1 += delta_wh1
        wh2 += delta_wh2

        b1 += delta_b1
        b2 += delta_b2
        b3 += delta_b3

    return w11, w12, w21, w22, wh1, wh2, b1, b2, b3

  def predict(self, weights, x1, x2):
    hidden1 = 1 / (1 + np.exp(-((x1 * weights[0]) + (x2 * weights[2]) + weights[6])))
    hidden2 = 1 / (1 + np.exp(-((x1 * weights[1]) + (x2 * weights[3]) + weights[7])))

    # Condição de retorno
    return 1 if 1 / (1 + np.exp(-((hidden1 * weights[4]) + (hidden2 * weights[5]) + weights[8]))) > 0.5 else 0
````

# Teste 1

````
# Dados de Entrada e Saída baseados na tabela da verdade da porta XOR
inputs = [[0,0], [0,1], [1,0], [1,1]]
outputs = [[0], [1], [1], [0]]

mlp = MLP()
# Seta os valores de treinamento -> (inputs, outputs, Taxa de Aprendizado, épocas)
tr = mlp.train(inputs, outputs, 0.1, 10000)

#Resultado do treinamento de acordo com os dados computados anteriormente
y = mlp.predict(tr, 1, 1)
print(y)
````

# Teste 2 — Tabela da verdade completa

No Teste 1 verificamos apenas a predição para a entrada `[1, 1]`. Aqui percorremos as 4 entradas possíveis da porta XOR para confirmar se a rede aprendeu **todos** os padrões, não apenas um.

````
# Dados de Entrada e Saída baseados na tabela da verdade da porta XOR
inputs  = [[0,0], [0,1], [1,0], [1,1]]
outputs = [[0],   [1],   [1],   [0]]

mlp2 = MLP()
# Seta os valores de treinamento -> (inputs, outputs, Taxa de Aprendizado, épocas)
tr2 = mlp2.train(inputs, outputs, 0.1, 10000)

#Percorre todas as combinações de entrada da porta XOR e exibe o resultado predito
#[Antes só verificávamos predict(tr, 1, 1) — aqui iteramos sobre todo o conjunto]
print(f"{'X1':>4} {'X2':>4} {'Esperado':>10} {'Predito':>9}")
print('-' * 32)
for (x1, x2), y_true in zip(inputs, outputs):
    y_pred = mlp2.predict(tr2, x1, x2)
    print(f"{x1:>4} {x2:>4} {y_true[0]:>10} {y_pred:>9}")
````

# Teste 3 — Curva de aprendizado

Para saber **se e quando** a rede convergiu, adicionamos `historico_loss` ao `train`: a cada época calculamos o MSE médio sobre os 4 exemplos e guardamos na lista. Isso não altera a lógica de treinamento — é apenas um registro paralelo.

> **MSE (Erro Quadrático Médio):** média dos quadrados dos erros de cada exemplo. Quanto menor, mais próximas as predições estão do esperado. Valores abaixo de `0.01` indicam convergência adequada para o XOR.

````
# A criação de uma classe permite não deixar as variáveis "despersas", com uma Classe, tudo que é pertencente a uma rede neural, fica nela
# EX: class MLP: | def train(self) pass | def preditct (self) pass, podendo ser criado uma instância da rede neural, logo: rede = MLP() | rede.train() | rede.predict()
class MLPComCurva:
  def __init__(self): #"__init__" é utilizado para iniciar atributos, nesse caso o ""self
    self.historico_loss = [] #[Lista para guardar o MSE de cada época — necessária para plotar a curva de aprendizado]

  def train(self, inputs, outputs, alpha, epochs):
    #Parâmetros que a função train deverá receber
    #[Aqui o self é essencial para chamar os parâmetros fora da função]
    self.inputs = inputs
    self.outputs = outputs
    self.alpha = alpha   #Taxa de aprendizado do Algoritmo
    self.epochs = epochs #Quantidade de Iterações na etapa de treinamento
    self.historico_loss = [] #[Reinicia o histórico a cada chamada de train, evitando acumular resultados de treinos anteriores]

    #Definição dos Pesos e Bias (inicialização aleaória)
    w11 = np.random.uniform(0,1) #[np(chama o numpy).random(seta aleatoriedade).uniform(seta uma aleatoriedade "uniform" entre 0 e 1)]
    w12 = np.random.uniform(0,1)
    w21 = np.random.uniform(0,1)
    w22 = np.random.uniform(0,1)

    wh1 = np.random.uniform(0,1)
    wh2 = np.random.uniform(0,1)

    b1 = np.random.uniform(0,1)
    b2 = np.random.uniform(0,1)
    b3 = np.random.uniform(0,1)

    #LAÇOS DE REPETIÇÃO PARA TREINAMENTO
    for i in range(epochs):         #Iteração para a quantidade de épocas selecionadas
      loss_epoca = 0.0              #[Acumulador de erro para calcular o MSE ao final de cada época]
      for j in range(len(inputs)):  #"len(inputs)" retorna quantos exemplos de treinamento existem considerando todo o conjunto de dados(inputs)
                                    # Por exemplo: inputs = {[0,0] [0,1] [1,0] [1,1]}

    #Execução de cálculos que serão passados adiantes para a camada oculta
    #Utiliza a função de ativação sigmoid, que possibilita transformar qualquer valor em um número entre 0 e 1 --> f(x) = 1 / 1 + e⁻x
    # Então aplicando a o sigmoid temos: h = f(x1*w1 + x2*w2 + b1 ) -> em que o resultado do "treinamento" é convertido para um valor entre 0 e 1
        h1 = 1 / (1 + np.exp(-(w11 * inputs[j][0] + w12 * inputs[j][1] + b1)))
        h2 = 1 / (1 + np.exp(-(w21 * inputs[j][0] + w22 * inputs[j][1] + b2)))

        #Execução de cálculos que serão passados adiantes para a camada de saída
        y = 1 / (1 + np.exp(-(wh1 * h1 + wh2 * h2 + b3)))

        #Calculamos o erro
        error = outputs[j][0] - y # Access the scalar value from the list
        loss_epoca += error ** 2  #[Soma o quadrado do erro deste exemplo — o MSE será a média ao final do loop interno]

        #Calculo de derivadas parciais
        derivative_y = y * (1 - y) * error
        derivative_h1 = h1 * (1 - h1) * wh1 * derivative_y
        derivative_h2 = h2 * (1 - h2) * wh2 * derivative_y

        #Calculo dos deltas que serão utilizados para atualização dos pesos
        delta_w11 = alpha * derivative_h1 * inputs[j][0]
        delta_w12 = alpha * derivative_h2 * inputs[j][0]
        delta_w21 = alpha * derivative_h1 * inputs[j][1]
        delta_w22 = alpha * derivative_h2 * inputs[j][1]

        delta_b1 = alpha * derivative_h1
        delta_b2 = alpha * derivative_h2
        delta_b3 = alpha * derivative_y

        delta_wh1 = alpha * derivative_y * h1
        delta_wh2 = alpha * derivative_y * h2

        #Atualização dos pesos
        w11 += delta_w11
        w12 += delta_w12
        w21 += delta_w21
        w22 += delta_w22

        wh1 += delta_wh1
        wh2 += delta_wh2

        b1 += delta_b1
        b2 += delta_b2
        b3 += delta_b3

      #[Após processar todos os exemplos da época, calcula o MSE médio e salva no histórico]
      self.historico_loss.append(loss_epoca / len(inputs))

    return w11, w12, w21, w22, wh1, wh2, b1, b2, b3

  def predict(self, weights, x1, x2):
    hidden1 = 1 / (1 + np.exp(-((x1 * weights[0]) + (x2 * weights[2]) + weights[6])))
    hidden2 = 1 / (1 + np.exp(-((x1 * weights[1]) + (x2 * weights[3]) + weights[7])))

    # Condição de retorno
    return 1 if 1 / (1 + np.exp(-((hidden1 * weights[4]) + (hidden2 * weights[5]) + weights[8]))) > 0.5 else 0
````

````
# Dados de Entrada e Saída baseados na tabela da verdade da porta XOR
inputs  = [[0,0], [0,1], [1,0], [1,1]]
outputs = [[0],   [1],   [1],   [0]]

mlp3 = MLPComCurva()
# Seta os valores de treinamento -> (inputs, outputs, Taxa de Aprendizado, épocas)
tr3 = mlp3.train(inputs, outputs, 0.1, 10000)

#Resultado do treinamento de acordo com os dados computados anteriormente
print(f"{'X1':>4} {'X2':>4} {'Esperado':>10} {'Predito':>9}")
print('-' * 32)
for (x1, x2), y_true in zip(inputs, outputs):
    y_pred = mlp3.predict(tr3, x1, x2)
    print(f"{x1:>4} {x2:>4} {y_true[0]:>10} {y_pred:>9}")

#Plot da curva de aprendizado
#[O eixo x representa as épocas e o eixo y o MSE — queremos ver uma queda consistente até abaixo de 0.01]
plt.figure(figsize=(8, 4))
plt.plot(mlp3.historico_loss, color='steelblue', linewidth=1.5, label='MSE treino')
plt.axhline(0.01, color='tomato', linestyle='--', linewidth=1, label='Meta (MSE = 0.01)')
plt.xlabel('Época')
plt.ylabel('MSE')
plt.title('Curva de aprendizado — alpha=0.1 | epochs=10000')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

print(f'\nMSE final: {mlp3.historico_loss[-1]:.6f}')
````

# Teste 4 — Efeito do alpha e do número de épocas

O `alpha` (taxa de aprendizado) e o número de épocas são os dois principais **hiperparâmetros** desta rede — valores definidos antes do treino, não aprendidos por ela.

- **alpha muito alto:** os pesos "saltam" demais a cada atualização e a loss oscila em vez de cair
- **alpha muito baixo:** a loss cai, mas devagar — pode precisar de muito mais épocas para convergir
- **épocas insuficientes:** a rede para de treinar antes de convergir

Aqui rodamos o mesmo treino com diferentes combinações e comparamos os resultados lado a lado.

````
# Dados de Entrada e Saída baseados na tabela da verdade da porta XOR
inputs  = [[0,0], [0,1], [1,0], [1,1]]
outputs = [[0],   [1],   [1],   [0]]

#Combinações de alpha e epochs a comparar
#[Cada tupla representa (alpha, epochs) — o objetivo é observar como cada par afeta a convergência]
experimentos = [
    (0.01, 10000),
    (0.1,  10000),  #Configuração do Teste 1 (ponto de referência)
    (0.3,  10000),
    (0.1,   3000),
    (0.1,  30000),
]

fig, axes = plt.subplots(1, len(experimentos), figsize=(18, 3.5))

print(f"{'alpha':>8}  {'epochs':>8}  {'MSE final':>10}  {'Acurácia':>10}")
print('-' * 44)

for ax, (a, e) in zip(axes, experimentos):
    r = MLPComCurva()
    tr_exp = r.train(inputs, outputs, a, e)

    #Calcula acurácia sobre os 4 padrões
    acertos = sum(r.predict(tr_exp, x[0], x[1]) == y[0] for x, y in zip(inputs, outputs))

    ax.plot(r.historico_loss, color='steelblue', linewidth=1)
    ax.axhline(0.01, color='tomato', linestyle='--', linewidth=0.8)
    ax.set_title(f'alpha={a} | ep={e}\nMSE={r.historico_loss[-1]:.4f} | acc={acertos}/4', fontsize=8)
    ax.set_xlabel('Época', fontsize=8)
    ax.set_ylabel('MSE',   fontsize=8)
    ax.grid(True, alpha=0.2)

    print(f"{a:>8}  {e:>8}  {r.historico_loss[-1]:>10.5f}  {acertos}/4")

plt.suptitle('Comparação de alpha × epochs', y=1.02)
plt.tight_layout()
plt.show()
````

# Teste 5 — MLP modular (N neurônios na camada oculta)

Nas versões anteriores a arquitetura estava **fixa**: exatamente 2 neurônios ocultos, com variáveis individuais (`w11`, `w12`, `w21`, `w22`…). Isso funciona para o XOR, mas não escala — no MNIST a camada de entrada terá 784 neurônios e a oculta pode ter 128 ou 256.

A solução é substituir as variáveis individuais por **matrizes de pesos** usando `np.random.uniform` sobre arrays:
- `W1` → matriz de pesos da camada oculta, shape `(n_ocultos, n_entradas)`
- `W2` → vetor de pesos da camada de saída, shape `(1, n_ocultos)`
- `b1`, `b2` → vetores de bias

O cálculo do sigmoid e do backprop permanece idêntico ao que já foi implementado — só a forma de armazenar os pesos muda.

````
# A criação de uma classe permite não deixar as variáveis "despersas", com uma Classe, tudo que é pertencente a uma rede neural, fica nela
# EX: class MLP: | def train(self) pass | def preditct (self) pass, podendo ser criado uma instância da rede neural, logo: rede = MLP() | rede.train() | rede.predict()
class MLPModular:
  def __init__(self, n_entradas, n_ocultos, n_saidas=1):
    #"__init__" é utilizado para iniciar atributos, nesse caso o ""self
    #[Recebe a arquitetura como parâmetro em vez de deixá-la fixa no código]
    #[EX: MLPModular(2, 2) → mesma arquitetura do Teste 1 | MLPModular(2, 4) → camada oculta com 4 neurônios]
    self.n_entradas = n_entradas
    self.n_ocultos  = n_ocultos
    self.n_saidas   = n_saidas
    self.historico_loss = []

  def train(self, inputs, outputs, alpha, epochs):
    #Parâmetros que a função train deverá receber
    #[Aqui o self é essencial para chamar os parâmetros fora da função]
    self.inputs  = inputs
    self.outputs = outputs
    self.alpha   = alpha   #Taxa de aprendizado do Algoritmo
    self.epochs  = epochs  #Quantidade de Iterações na etapa de treinamento
    self.historico_loss = []

    #Definição dos Pesos e Bias (inicialização aleaória)
    #[Substituímos as variáveis individuais (w11, w12...) por matrizes — isso permite controlar
    # a quantidade de neurônios sem reescrever o código de treino]
    #[W1 shape: (n_ocultos, n_entradas) — cada linha é um neurônio oculto, cada coluna é uma entrada]
    #[np(chama o numpy).random(seta aleatoriedade).uniform(seta uma aleatoriedade "uniform" entre 0 e 1)]
    W1 = np.random.uniform(0, 1, (self.n_ocultos, self.n_entradas))
    b1 = np.random.uniform(0, 1, (self.n_ocultos, 1))

    #[W2 shape: (n_saidas, n_ocultos) — pesos que conectam a camada oculta à saída]
    W2 = np.random.uniform(0, 1, (self.n_saidas, self.n_ocultos))
    b2 = np.random.uniform(0, 1, (self.n_saidas, 1))

    #LAÇOS DE REPETIÇÃO PARA TREINAMENTO
    for i in range(epochs):           #Iteração para a quantidade de épocas selecionadas
      loss_epoca = 0.0
      for j in range(len(inputs)):    #"len(inputs)" retorna quantos exemplos de treinamento existem considerando todo o conjunto de dados(inputs)
                                      # Por exemplo: inputs = {[0,0] [0,1] [1,0] [1,1]}

        #[Converte o exemplo j para vetor coluna shape (n_entradas, 1) — necessário para a multiplicação matricial]
        x = np.array(inputs[j]).reshape(-1, 1)
        y_true = outputs[j][0]

        #Execução de cálculos que serão passados adiantes para a camada oculta
        #Utiliza a função de ativação sigmoid, que possibilita transformar qualquer valor em um número entre 0 e 1 --> f(x) = 1 / 1 + e⁻x
        # Então aplicando a o sigmoid temos: h = f(W1*x + b1) -> em que o resultado do "treinamento" é convertido para um valor entre 0 e 1
        #[W1 @ x realiza a mesma soma ponderada de antes: cada neurônio oculto recebe todas as entradas]
        z1 = W1 @ x + b1
        h  = 1 / (1 + np.exp(-z1))   #Sigmoid aplicado elemento a elemento sobre o vetor z1

        #Execução de cálculos que serão passados adiantes para a camada de saída
        z2 = W2 @ h + b2
        y  = 1 / (1 + np.exp(-z2))   #Sigmoid aplicado à saída

        #Calculamos o erro
        error = y_true - y            # Access the scalar value from the list
        loss_epoca += float(error ** 2)

        #Calculo de derivadas parciais
        #[A estrutura do backprop é idêntica à versão original — só trocamos variáveis escalares por vetores]
        derivative_y = y  * (1 - y)  * error          #Gradiente na saída
        derivative_h = h  * (1 - h)  * (W2.T @ derivative_y)  #Gradiente propagado para a camada oculta

        #Calculo dos deltas que serão utilizados para atualização dos pesos
        delta_W2 = alpha * derivative_y @ h.T
        delta_b2 = alpha * derivative_y

        delta_W1 = alpha * derivative_h @ x.T
        delta_b1 = alpha * derivative_h

        #Atualização dos pesos
        W2 += delta_W2
        b2 += delta_b2
        W1 += delta_W1
        b1 += delta_b1

      #[Após processar todos os exemplos da época, calcula o MSE médio e salva no histórico]
      self.historico_loss.append(loss_epoca / len(inputs))

    #[Salva os pesos finais como atributos para uso no predict]
    self.W1 = W1; self.b1 = b1
    self.W2 = W2; self.b2 = b2

  def predict(self, x1, x2):
    #[predict agora usa os pesos salvos em self — não precisa receber a tupla de pesos como parâmetro]
    x  = np.array([x1, x2]).reshape(-1, 1)
    h  = 1 / (1 + np.exp(-(self.W1 @ x + self.b1)))
    y  = 1 / (1 + np.exp(-(self.W2 @ h + self.b2)))

    # Condição de retorno
    return 1 if float(y) > 0.5 else 0
````

````
# Dados de Entrada e Saída baseados na tabela da verdade da porta XOR
inputs  = [[0,0], [0,1], [1,0], [1,1]]
outputs = [[0],   [1],   [1],   [0]]

#Testa com a mesma arquitetura do Teste 1: 2 entradas, 2 neurônios ocultos, 1 saída
#[Passando n_ocultos=2 reproduzimos exatamente o que w11/w12/w21/w22 representavam antes]
mlp5 = MLPModular(n_entradas=2, n_ocultos=2)
# Seta os valores de treinamento -> (inputs, outputs, Taxa de Aprendizado, épocas)
mlp5.train(inputs, outputs, alpha=0.1, epochs=10000)

#Resultado do treinamento de acordo com os dados computados anteriormente
print(f"{'X1':>4} {'X2':>4} {'Esperado':>10} {'Predito':>9}")
print('-' * 32)
for (x1, x2), y_true in zip(inputs, outputs):
    y_pred = mlp5.predict(x1, x2)
    print(f"{x1:>4} {x2:>4} {y_true[0]:>10} {y_pred:>9}")

#Plot da curva de aprendizado
#[O eixo x representa as épocas e o eixo y o MSE — queremos ver uma queda consistente até abaixo de 0.01]
plt.figure(figsize=(8, 4))
plt.plot(mlp5.historico_loss, color='steelblue', linewidth=1.5, label='MSE treino')
plt.axhline(0.01, color='tomato', linestyle='--', linewidth=1, label='Meta (MSE = 0.01)')
plt.xlabel('Época')
plt.ylabel('MSE')
plt.title('Curva de aprendizado — MLPModular(2 entradas, 2 ocultos) | alpha=0.1 | epochs=10000')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

print(f'\nMSE final: {mlp5.historico_loss[-1]:.6f}')
````

# Teste 6 — Efeito do número de neurônios ocultos

Com a `MLPModular` é possível variar `n_ocultos` sem reescrever o treino. Para o XOR — um problema simples — 2 neurônios já são suficientes. O experimento abaixo mostra o que acontece com 1, 2, 4 e 8 neurônios:

- **1 neurônio:** capacidade insuficiente — o XOR não é linearmente separável e um único neurônio oculto pode não conseguir aprendê-lo
- **2 neurônios:** mínimo funcional para o XOR
- **4 / 8 neurônios:** mais capacidade do que o necessário — a rede ainda converge, mas os neurônios extras ficam redundantes

> No MNIST essa escolha importa muito mais: poucos neurônios limitam o que a rede consegue aprender; neurônios demais aumentam o tempo de treino e o risco de overfitting.

````
# Dados de Entrada e Saída baseados na tabela da verdade da porta XOR
inputs  = [[0,0], [0,1], [1,0], [1,1]]
outputs = [[0],   [1],   [1],   [0]]

#Variações de neurônios ocultos a comparar
#[Apenas n_ocultos muda — alpha e epochs ficam fixos para isolar o efeito da arquitetura]
variacoes_ocultos = [1, 2, 4, 8]

fig, axes = plt.subplots(1, len(variacoes_ocultos), figsize=(14, 3.5))

print(f"{'n_ocultos':>10}  {'MSE final':>10}  {'Acurácia':>10}")
print('-' * 36)

for ax, n in zip(axes, variacoes_ocultos):
    r = MLPModular(n_entradas=2, n_ocultos=n)
    r.train(inputs, outputs, alpha=0.1, epochs=10000)

    #Calcula acurácia sobre os 4 padrões
    acertos = sum(r.predict(x[0], x[1]) == y[0] for x, y in zip(inputs, outputs))

    ax.plot(r.historico_loss, color='darkorchid', linewidth=1)
    ax.axhline(0.01, color='tomato', linestyle='--', linewidth=0.8)
    ax.set_title(f'n_ocultos={n}\nMSE={r.historico_loss[-1]:.4f} | acc={acertos}/4', fontsize=9)
    ax.set_xlabel('Época', fontsize=8)
    ax.set_ylabel('MSE',   fontsize=8)
    ax.grid(True, alpha=0.2)

    print(f"{n:>10}  {r.historico_loss[-1]:>10.5f}  {acertos}/4")

plt.suptitle('Efeito do número de neurônios ocultos — alpha=0.1 | epochs=10000', y=1.02)
plt.tight_layout()
plt.show()
````

# Teste 7 — MLP com N camadas ocultas

A `MLPModular` do Teste 5 suporta exatamente **uma** camada oculta. Para a atividade ponderada o requisito mínimo é **duas** camadas ocultas. Aqui a arquitetura é generalizada para aceitar uma lista de tamanhos de camadas.

A mudança estrutural é substituir os pares fixos `W1/b1` e `W2/b2` por uma **lista** de matrizes `self.pesos` e `self.biases`. O forward pass percorre essa lista em sequência; o backpropagation percorre em sentido inverso.

Essa é a versão que será diretamente reaplicada no MNIST.

````
# A criação de uma classe permite não deixar as variáveis "despersas", com uma Classe, tudo que é pertencente a uma rede neural, fica nela
# EX: class MLP: | def train(self) pass | def preditct (self) pass, podendo ser criado uma instância da rede neural, logo: rede = MLP() | rede.train() | rede.predict()
class MLPProfunda:
  def __init__(self, camadas):
    #"__init__" é utilizado para iniciar atributos, nesse caso o "self"
    #[camadas é uma lista com o tamanho de cada camada, incluindo entrada e saída]
    #[EX: MLPProfunda([2, 4, 4, 1]) → entrada com 2, duas ocultas com 4 neurônios cada, saída com 1]
    self.camadas = camadas
    self.historico_loss = []
    self._inicializar_pesos()

  def _inicializar_pesos(self):
    #[Cria uma lista de matrizes W e vetores b, um par por camada de conexão]
    #[Para uma rede [2, 4, 4, 1]: há 3 pares (W, b) — camada 0→1, 1→2, 2→3]
    self.pesos  = []
    self.biases = []
    for i in range(len(self.camadas) - 1):
      n_entrada = self.camadas[i]
      n_saida   = self.camadas[i + 1]
      #[np(chama o numpy).random(seta aleatoriedade).uniform(seta uma aleatoriedade "uniform" entre -1 e 1)]
      #[Usando uniform(-1,1) em vez de uniform(0,1) para evitar viés positivo que satura a sigmoid]
      self.pesos.append(np.random.uniform(-1, 1, (n_saida, n_entrada)))
      self.biases.append(np.random.uniform(-1, 1, (n_saida, 1)))

  def _sigmoid(self, z):
    #Utiliza a função de ativação sigmoid, que possibilita transformar qualquer valor em um número entre 0 e 1 --> f(x) = 1 / 1 + e⁻x
    return 1 / (1 + np.exp(-z))

  def _sigmoid_deriv(self, s):
    #[Derivada da sigmoid em função do valor já ativado: s*(1-s)]
    return s * (1 - s)

  def _forward(self, x):
    #[Percorre a lista de pesos em sequência — cada camada recebe a saída da anterior]
    #[Guarda todas as ativações intermediárias pois o backprop vai precisar delas]
    ativacoes = [x]
    for W, b in zip(self.pesos, self.biases):
      z = W @ ativacoes[-1] + b
      ativacoes.append(self._sigmoid(z))
    return ativacoes

  def train(self, inputs, outputs, alpha, epochs):
    #Parâmetros que a função train deverá receber
    #[Aqui o self é essencial para chamar os parâmetros fora da função]
    self.alpha  = alpha   #Taxa de aprendizado do Algoritmo
    self.epochs = epochs  #Quantidade de Iterações na etapa de treinamento
    self.historico_loss = []

    #LAÇOS DE REPETIÇÃO PARA TREINAMENTO
    for i in range(epochs):         #Iteração para a quantidade de épocas selecionadas
      loss_epoca = 0.0              #[Acumulador de erro para calcular o MSE ao final de cada época]
      for j in range(len(inputs)):  #"len(inputs)" retorna quantos exemplos de treinamento existem considerando todo o conjunto de dados(inputs)
                                    # Por exemplo: inputs = {[0,0] [0,1] [1,0] [1,1]}
        #[Converte o exemplo j para vetor coluna shape (n_entradas, 1) — necessário para a multiplicação matricial]
        x      = np.array(inputs[j]).reshape(-1, 1)
        y_true = outputs[j][0]

        #Execução de cálculos que serão passados adiantes (forward pass)
        ativacoes = self._forward(x)
        y = float(ativacoes[-1])

        #Calculamos o erro
        error = y_true - y
        loss_epoca += error ** 2

        #Calculo de derivadas parciais — backpropagation percorrendo a lista de camadas ao contrário
        #[delta_saida: gradiente na camada de saída — mesma fórmula dos testes anteriores]
        delta = self._sigmoid_deriv(ativacoes[-1]) * error

        #[Percorre as camadas de trás pra frente, propagando o gradiente]
        for k in range(len(self.pesos) - 1, -1, -1):
          #Calculo dos deltas que serão utilizados para atualização dos pesos
          dW = alpha * delta @ ativacoes[k].T
          db = alpha * delta

          #[Propaga o gradiente para a camada anterior antes de atualizar os pesos]
          if k > 0:
            delta = self._sigmoid_deriv(ativacoes[k]) * (self.pesos[k].T @ delta)

          #Atualização dos pesos
          self.pesos[k]  += dW
          self.biases[k] += db

      #[Após processar todos os exemplos da época, calcula o MSE médio e salva no histórico]
      self.historico_loss.append(loss_epoca / len(inputs))

  def predict(self, x1, x2):
    #[predict agora usa os pesos salvos em self — não precisa receber a tupla de pesos como parâmetro]
    x = np.array([x1, x2]).reshape(-1, 1)
    ativacoes = self._forward(x)

    # Condição de retorno
    return 1 if float(ativacoes[-1]) > 0.5 else 0
````

````
# Dados de Entrada e Saída baseados na tabela da verdade da porta XOR
inputs  = [[0,0], [0,1], [1,0], [1,1]]
outputs = [[0],   [1],   [1],   [0]]

#Testa com arquitetura de duas camadas ocultas — mesma exigência da atividade ponderada
#[2 entradas → 4 neurônios ocultos → 4 neurônios ocultos → 1 saída]
mlp7 = MLPProfunda([2, 4, 4, 1])
mlp7.train(inputs, outputs, alpha=0.1, epochs=10000)

#Resultado do treinamento de acordo com os dados computados anteriormente
print(f"{'X1':>4} {'X2':>4} {'Esperado':>10} {'Predito':>9}")
print('-' * 32)
for (x1, x2), y_true in zip(inputs, outputs):
    y_pred = mlp7.predict(x1, x2)
    print(f"{x1:>4} {x2:>4} {y_true[0]:>10} {y_pred:>9}")

#Plot da curva de aprendizado
#[O eixo x representa as épocas e o eixo y o MSE — queremos ver uma queda consistente até abaixo de 0.01]
plt.figure(figsize=(8, 4))
plt.plot(mlp7.historico_loss, color='steelblue', linewidth=1.5, label='MSE treino')
plt.axhline(0.01, color='tomato', linestyle='--', linewidth=1, label='Meta (MSE = 0.01)')
plt.xlabel('Época')
plt.ylabel('MSE')
plt.title('Curva de aprendizado — MLPProfunda([2,4,4,1]) | alpha=0.1 | epochs=10000')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

print(f'\nMSE final: {mlp7.historico_loss[-1]:.6f}')
````

# Teste 8 — Comparação sigmoid vs ReLU

Até aqui todas as camadas ocultas usam **sigmoid** como função de ativação. A atividade ponderada exige **ReLU** como ativação configurável nas camadas ocultas.

**ReLU (Rectified Linear Unit):**

```
f(z) = max(0, z)
```

Sua derivada é:

```
f'(z) = 1 se z > 0, senão 0
```

A sigmoid tem gradiente próximo de zero para valores extremos. Em redes profundas isso faz o sinal de erro encolher ao ser propagado para trás por várias camadas (**vanishing gradient**). A ReLU não sofre desse problema para valores positivos: sua derivada é 1 e o gradiente não encolhe. Para o XOR a diferença é pequena. Para o MNIST com duas camadas ocultas, a ReLU converge mais rápido e de forma mais estável.

````
# A criação de uma classe permite não deixar as variáveis "despersas", com uma Classe, tudo que é pertencente a uma rede neural, fica nela
# EX: class MLP: | def train(self) pass | def preditct (self) pass, podendo ser criado uma instância da rede neural, logo: rede = MLP() | rede.train() | rede.predict()
class MLPAtivacaoConfig:
  def __init__(self, camadas, ativacao='sigmoid'):
    #"__init__" é utilizado para iniciar atributos, nesse caso o "self"
    #[ativacao define qual função será usada nas camadas ocultas — 'sigmoid' ou 'relu']
    #[A camada de saída sempre usa sigmoid nesta versão (saída binária para o XOR)]
    self.camadas   = camadas
    self.ativacao  = ativacao
    self.historico_loss = []
    self._inicializar_pesos()

  def _inicializar_pesos(self):
    #[Cria uma lista de matrizes W e vetores b, um par por camada de conexão]
    self.pesos  = []
    self.biases = []
    for i in range(len(self.camadas) - 1):
      n_entrada = self.camadas[i]
      n_saida   = self.camadas[i + 1]
      #[np(chama o numpy).random(seta aleatoriedade).uniform(seta uma aleatoriedade "uniform" entre -1 e 1)]
      self.pesos.append(np.random.uniform(-1, 1, (n_saida, n_entrada)))
      self.biases.append(np.random.uniform(-1, 1, (n_saida, 1)))

  def _ativar(self, z, saida=False):
    #[saida=True força sigmoid independente da ativação configurada — a camada de saída do XOR é sempre sigmoid]
    if saida or self.ativacao == 'sigmoid':
      return 1 / (1 + np.exp(-z))
    #[ReLU: max(0, z) aplicado elemento a elemento — np.maximum compara cada elemento de z com 0]
    return np.maximum(0, z)

  def _ativar_deriv(self, a, saida=False):
    #[Derivada da função de ativação em função do valor já ativado]
    if saida or self.ativacao == 'sigmoid':
      #[Derivada da sigmoid: s*(1-s)]
      return a * (1 - a)
    #[Derivada da ReLU: 1 onde a > 0, 0 onde a == 0]
    return (a > 0).astype(float)

  def _forward(self, x):
    #[Percorre a lista de pesos em sequência — cada camada recebe a saída da anterior]
    #[Guarda todas as ativações intermediárias pois o backprop vai precisar delas]
    ativacoes = [x]
    n_camadas = len(self.pesos)
    for idx, (W, b) in enumerate(zip(self.pesos, self.biases)):
      z = W @ ativacoes[-1] + b
      #[A última camada (saída) sempre usa sigmoid; as ocultas usam a ativação configurada]
      eh_saida = (idx == n_camadas - 1)
      ativacoes.append(self._ativar(z, saida=eh_saida))
    return ativacoes

  def train(self, inputs, outputs, alpha, epochs):
    #Parâmetros que a função train deverá receber
    #[Aqui o self é essencial para chamar os parâmetros fora da função]
    self.historico_loss = []
    n_camadas = len(self.pesos)

    #LAÇOS DE REPETIÇÃO PARA TREINAMENTO
    for i in range(epochs):         #Iteração para a quantidade de épocas selecionadas
      loss_epoca = 0.0
      for j in range(len(inputs)):
        x      = np.array(inputs[j]).reshape(-1, 1)
        y_true = outputs[j][0]

        #Execução de cálculos que serão passados adiantes (forward pass)
        ativacoes = self._forward(x)
        y = float(ativacoes[-1])

        #Calculamos o erro
        error = y_true - y
        loss_epoca += error ** 2

        #Calculo de derivadas parciais — backpropagation percorrendo a lista de camadas ao contrário
        delta = self._ativar_deriv(ativacoes[-1], saida=True) * error

        for k in range(n_camadas - 1, -1, -1):
          dW = alpha * delta @ ativacoes[k].T
          db = alpha * delta
          if k > 0:
            eh_saida_k = (k == n_camadas - 1)
            delta = self._ativar_deriv(ativacoes[k], saida=eh_saida_k) * (self.pesos[k].T @ delta)

          #Atualização dos pesos
          self.pesos[k]  += dW
          self.biases[k] += db

      #[Após processar todos os exemplos da época, calcula o MSE médio e salva no histórico]
      self.historico_loss.append(loss_epoca / len(inputs))

  def predict(self, x1, x2):
    #[predict agora usa os pesos salvos em self — não precisa receber a tupla de pesos como parâmetro]
    x = np.array([x1, x2]).reshape(-1, 1)
    ativacoes = self._forward(x)

    # Condição de retorno
    return 1 if float(ativacoes[-1]) > 0.5 else 0
````

````
# Dados de Entrada e Saída baseados na tabela da verdade da porta XOR
inputs  = [[0,0], [0,1], [1,0], [1,1]]
outputs = [[0],   [1],   [1],   [0]]

#Comparação entre sigmoid e ReLU na camada oculta
#[Ambas as redes têm a mesma arquitetura [2,4,1] e os mesmos hiperparâmetros — só a ativação muda]
configs_ativ = [
    ('sigmoid', 'steelblue'),
    ('relu',    'darkorchid'),
]

fig, axes = plt.subplots(1, 2, figsize=(12, 4))

print(f"{'Ativacao':>10}  {'MSE final':>10}  {'Acuracia':>10}")
print('-' * 36)

for ax, (ativ, cor) in zip(axes, configs_ativ):
    r = MLPAtivacaoConfig([2, 4, 1], ativacao=ativ)
    r.train(inputs, outputs, alpha=0.1, epochs=10000)

    #Calcula acurácia sobre os 4 padrões
    acertos = sum(r.predict(x[0], x[1]) == y[0] for x, y in zip(inputs, outputs))

    ax.plot(r.historico_loss, color=cor, linewidth=1.5, label=f'MSE ({ativ})')
    ax.axhline(0.01, color='tomato', linestyle='--', linewidth=0.8, label='Meta')
    ax.set_title(f'Ativacao: {ativ}\nMSE={r.historico_loss[-1]:.4f} | acc={acertos}/4', fontsize=9)
    ax.set_xlabel('Epoca')
    ax.set_ylabel('MSE')
    ax.legend(fontsize=8)
    ax.grid(True, alpha=0.3)

    print(f"{ativ:>10}  {r.historico_loss[-1]:>10.5f}  {acertos}/4")

plt.suptitle('Sigmoid vs ReLU — arquitetura [2,4,1] | alpha=0.1 | epochs=10000')
plt.tight_layout()
plt.show()
````

# Teste 9 — Inicialização de pesos

Nos testes anteriores os pesos foram inicializados com `np.random.uniform(0, 1)`. Esse esquema tem um problema: todos os valores são positivos, o que gera ativações inicialmente altas na sigmoid e gradientes próximos de zero desde o começo do treinamento.

Três esquemas são comparados:

**`uniform(0, 1)`** — inicialização dos primeiros testes. Viés positivo.

**`uniform(-1, 1)`** — elimina o viés positivo. Mais equilibrado, mas sem relação com o tamanho da camada.

**Xavier (Glorot)** — inicializa os pesos em `[-√(1/n), √(1/n)]`, onde `n` é o número de entradas da camada. Projetado para manter a variância dos gradientes estável ao longo das camadas.

```python
limite = np.sqrt(1 / n_entradas)
W = np.random.uniform(-limite, limite, shape)
```

> **Conexão com o MNIST:** no MNIST com duas camadas ocultas a inicialização tem impacto direto na velocidade de convergência. A inicialização adotada na ponderada será Xavier.

````
# Dados de Entrada e Saída baseados na tabela da verdade da porta XOR
inputs  = [[0,0], [0,1], [1,0], [1,1]]
outputs = [[0],   [1],   [1],   [0]]

def treinar_com_init(esquema, inputs, outputs, alpha=0.1, epochs=10000):
    #[Reutiliza MLPAtivacaoConfig mas substitui a inicialização conforme o esquema]
    r = MLPAtivacaoConfig([2, 4, 1], ativacao='sigmoid')
    for i, (n_ent, n_sai) in enumerate(zip([2, 4], [4, 1])):
        if esquema == 'pos':
            #[uniform(0,1): inicialização dos primeiros testes — viés positivo]
            r.pesos[i]  = np.random.uniform(0,  1,  (n_sai, n_ent))
            r.biases[i] = np.random.uniform(0,  1,  (n_sai, 1))
        elif esquema == 'simetrico':
            #[uniform(-1,1): elimina o viés positivo]
            r.pesos[i]  = np.random.uniform(-1, 1,  (n_sai, n_ent))
            r.biases[i] = np.random.uniform(-1, 1,  (n_sai, 1))
        elif esquema == 'xavier':
            #[Xavier: escala os pesos pelo número de entradas da camada para manter a variância dos gradientes]
            limite = np.sqrt(1 / n_ent)
            r.pesos[i]  = np.random.uniform(-limite, limite, (n_sai, n_ent))
            r.biases[i] = np.zeros((n_sai, 1))
    r.train(inputs, outputs, alpha, epochs)
    return r

#Comparação dos três esquemas de inicialização
esquemas_init = [
    ('pos',       'uniform(0,1)',  'steelblue'),
    ('simetrico', 'uniform(-1,1)', 'darkorchid'),
    ('xavier',    'Xavier',        'seagreen'),
]

fig, axes = plt.subplots(1, 3, figsize=(15, 4))

print(f"{'Esquema':>15}  {'MSE final':>10}  {'Acuracia':>10}")
print('-' * 42)

for ax, (esq, label, cor) in zip(axes, esquemas_init):
    r = treinar_com_init(esq, inputs, outputs)

    #Calcula acurácia sobre os 4 padrões
    acertos = sum(r.predict(x[0], x[1]) == y[0] for x, y in zip(inputs, outputs))

    ax.plot(r.historico_loss, color=cor, linewidth=1.5)
    ax.axhline(0.01, color='tomato', linestyle='--', linewidth=0.8, label='Meta')
    ax.set_title(f'{label}\nMSE={r.historico_loss[-1]:.4f} | acc={acertos}/4', fontsize=9)
    ax.set_xlabel('Epoca')
    ax.set_ylabel('MSE')
    ax.legend(fontsize=8)
    ax.grid(True, alpha=0.3)

    print(f"{label:>15}  {r.historico_loss[-1]:>10.5f}  {acertos}/4")

plt.suptitle('Efeito da inicializacao de pesos — [2,4,1] | sigmoid | alpha=0.1 | epochs=10000')
plt.tight_layout()
plt.show()
````
